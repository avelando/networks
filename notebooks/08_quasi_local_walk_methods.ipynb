```python
from functools import partial

from link_prediction.config import (
    load_methods_config,
)
from link_prediction.evaluation import (
    run_method_family_benchmark,
)
from link_prediction.methods.quasi_local_walks import (
    score_quasi_local_walk_candidates,
)
```

```python
methods_config = (
    load_methods_config()[
        "methods"
    ]
)

walk_method_ids = [
    "lrw",
    "srw",
    "pfp",
]

configured_steps = {
    methods_config[
        method_id
    ][
        "parameters"
    ][
        "steps"
    ]
    for method_id
    in walk_method_ids
}

if len(
    configured_steps
) != 1:
    raise ValueError(
        "LRW, SRW, and PFP must use "
        "the same primary step configuration."
    )

steps = configured_steps.pop()

walk_scorer = partial(
    score_quasi_local_walk_candidates,
    steps=steps,
)

fold_metrics, network_summary = (
    run_method_family_benchmark(
        family_id="quasi_local_walk",
        scorer=walk_scorer,
        benchmark_name="revision",
    )
)

fold_metrics
```
