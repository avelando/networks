```python
from link_prediction.folds import build_benchmark_folds
```

```python
fold_summary = build_benchmark_folds(
    benchmark_name="standard",
)

fold_summary
```

```python
validation = (
    fold_summary
    .groupby(
        [
            "network_id",
            "network",
        ]
    )
    .agg(
        folds=("fold", "count"),
        positives=("positive_edges", "sum"),
        negatives=("negative_edges", "sum"),
        candidates=("candidate_pairs", "sum"),
        all_connected=(
            "training_connected",
            "all",
        ),
    )
)

validation
```
